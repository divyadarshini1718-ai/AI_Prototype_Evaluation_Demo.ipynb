## 1. Problem Statement

This project aims to develop an AI-based solution for solving a defined user problem.
The system accepts user input and generates an AI-based response.
The goal is to provide a simple and useful solution using an AI model or API.


## 2. AI Model / API Setup

```python
!pip install -q openai
```

```python
from openai import OpenAI
import os
from getpass import getpass

api_key = getpass("Enter your OpenAI API key: ")

client = OpenAI(api_key=api_key)

print("AI model setup completed successfully.")
```

## 3. Basic Prototype

```python
def get_ai_response(question):
      if not question.strip():
              return "Please enter a question."
      response = client.responses.create(
          model="gpt-5-mini",
          input=f"""
          You are a helpful AI Study Assistant.
          Explain concepts in simple English.
          Give short and clear answers suitable for students.
           Student question:
           {question}
           """
        )
        return response.output_text
        question = input("Enter your question: ")

        answer = get_ai_response(question)
        print("\nAI Response:")
        print(answer)
```

## 4. Example Inputs and Outputs

```python
test_questions = [
      "What is artificial intelligence?",
      "What is machine learning?",
      "What is a computer network?"
]

for question in test_questions:
                  print("=" * 60)
                  print("INPUT:", question)

                  answer = get_ai_response(question)

                  print("OUTPUT:")
                  print(answer)
```

## 5. Model Evaluation

The AI Study Assistant was evaluated using different types of questions.

The evaluation checks whether the AI provides clear, relevant, and
understandable answers.

Evaluation criteria:

1. Relevance – Does the answer address the question?
2. Clarity – Is the explanation easy to understand?
3. Completeness – Does the answer contain the important information?
4. Handling invalid input – Does the system respond appropriately?

```python
evaluation_questions = [
      "What is Python?",
      "Explain cloud computing.",
       "What is a database?",
       "",
       "Explain something in simple words."
]
for i, question in enumerate(evaluation_questions, 1):
      print(f"\nTest Case {i}")
      print("Input:", question)
      result = get_ai_response(question)

      print("Output:", result)
```

## 6. Failure Cases

## Failure Cases

The prototype may have limitations in the following situations:

1. Empty input:
   The system asks the user to enter a question.

   2. Unclear question:
      The AI may ask for clarification or provide a less precise answer.

      3. Unsupported question:
         The AI may not provide the expected answer if the question is
            outside the purpose of the study assistant.

            4. API or internet failure:
               The system may not generate a response if the AI API is unavailable.

```python
failure_cases = [
      "",
      "asdfghjkl",
       "???"
]
for question in failure_cases:
      print("=" * 60)
      print("TEST INPUT:", repr(question))

      try:
          result = get_ai_response(question)
          print("RESULT:", result)
      except Exception as error:
          print("ERROR:", error)
```

## 7. Simple Interface

```python
!pip install -q gradio
```

```python
import gradio as gr

def study_assistant(question):
    try:
        return get_ai_response(question)
    except Exception as error:
         return f"An error occurred: {error}"
    demo = gr.Interface(
        fn=study_assistant,
        inputs=gr.Textbox(
             label="Enter your question",
             placeholder="Example: What is machine learning?"
         ),
         outputs=gr.Textbox(
             label="AI Response"
         ),
         title="AI Study Assistant",
         description="Ask a question and get a simple AI-generated explanation."
    )
    demo.launch()
```
